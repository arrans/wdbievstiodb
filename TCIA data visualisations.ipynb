### Clean & Transfrom data

```python
# map cancer stages from codes to more meaningful text

cancer_stage = {
    110: "Stage 1A",
    120: "Stage 1B",
    210: "Stage 2A",
    220: "Stage 2B",
    310: "Stage 3A",
    320: "Stage 3B",
    400: "Stage 4",
    888: "TNM not available",
    900: "Occult Carcinoma",
    994: "Carcinoid, cannot be assessed",
    999: "Unknown, cannot be assessed",
    ".M": "Missing",
    ".N": "Not Applicable"
}

df_prsn['cancer_stage'] = (
    df_prsn['de_stag']
        .map(cancer_stage)
        .fillna('Other')
)

# create another layer of just the 4 stages
stage_group_map = {
    110: "Stage 1", 120: "Stage 1",
    210: "Stage 2", 220: "Stage 2",
    310: "Stage 3", 320: "Stage 3",
    400: "Stage 4"
}

df_prsn['cancer_stage_grouped'] = df_prsn['de_stag'].map(stage_group_map).fillna('Other')
```

```python
# map race code variable to meaningful text values
race_map = {
    1: "White",
    2: "Black or African-American",
    3: "Asian",
    4: "American Indian or Alaskan Native",
    5: "Native Hawaiian or Other Pacific Islander",
    6: "More than one race",
    7: "Participant refused to answer",
    95: "Missing data form - form is not expected to ever be completed",
    96: "Missing - no response",
    98: "Missing - form was submitted and the answer was left blank",
    99: "Unknown / decline to answer"
}

df_prsn['race_label'] = df_prsn['race'].map(race_map)
```

```python
# as others are insufficient in sample size to draw conclusions from, group them into other
race_map_grouped = {
    1: "White",
    2: "Black or African-American",
    3: "Asian"
}

df_prsn['race_grouped'] = (
    df_prsn['race']
        .map(race_map_grouped)
        .fillna('Other')
)
```

```python
# add age_group column

bins = [55, 60, 65, 70, 75, 80]
labels = ['55–59', '60–64', '65–69', '70–74', '75–79']

df_prsn['age_group'] = pd.cut(
    df_prsn['age'],
    bins=bins,
    labels=labels,
    right=False   # includes left edge (55–59 means 55 ≤ age < 60)
)
```

```python
# create new subset of data df_prsn_cancer_stg_1to4 made up of cancer related entries
```

```python
# for relevant analysis only keep entries which have cancer stages 
keep_stages = [110, 120, 210, 220, 310, 320, 400]
# df_plot = df_prsn[df_prsn['de_stag'].isin(keep_stages)]
df_prsn_cancer_stg_1to4 = df_prsn[df_prsn['de_stag'].isin(keep_stages)].copy()
cancer_stage_order = ["Stage 1A","Stage 1B","Stage 2A","Stage 2B","Stage 3A","Stage 3B","Stage 4","Other"]
```

```python
bins = [55, 60, 65, 70, 75, 80]
labels = ['55–59', '60–64', '65–69', '70–74', '75–79']

df_prsn_cancer_stg_1to4['age_group'] = pd.cut(
    df_prsn_cancer_stg_1to4['age'],
    bins=bins,
    labels=labels,
    right=False   # includes left edge (55–59 means 55 ≤ age < 60)
)
```

```python
# map gender labels
gender_labels = {1: 'Male', 2: 'Female'}

df_prsn['gender_label'] = df_prsn['gender'].map(gender_labels)
df_prsn_cancer_stg_1to4['gender_label'] = df_prsn_cancer_stg_1to4['gender'].map(gender_labels)
```
